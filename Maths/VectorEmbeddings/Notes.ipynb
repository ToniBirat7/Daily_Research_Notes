### **How Do Neural Networks Create Embeddings?**

`Neural networks`, including large language models like GPT-4, Llama-2, and Mistral-7B, create embeddings through a process called **representation learning**.

In this process, the network learns to map `high-dimensional` data into `lower-dimensional` spaces while preserving important properties of the data. They take raw input data, like images and texts, and represent them as numerical vectors.
